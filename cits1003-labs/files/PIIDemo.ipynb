```python
!pip install transformers

from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline

# Load the model and tokenizer
model_name = "TensaZangetsu/condensed-bert-vulnerable"
model = AutoModelForCausalLM.from_pretrained(model_name)
tokenizer = AutoTokenizer.from_pretrained(model_name)

# Create a text generation pipeline
generator = pipeline("text-generation", model=model, tokenizer=tokenizer)
```

```python
# Example of how to use the pipeline (demonstrating potential for data extraction)
text = "The capital of France is an"
output = generator(text, max_length=50, num_return_sequences=1)

print(f"Input text: {text}")
print(f"Generated text: {output[0]['generated_text']}")
```
